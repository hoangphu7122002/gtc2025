```python
len_ds = len(os.listdir('/workspace/nemo_datasets/add_id'))
for i in tqdm(range(len_ds)):
    if f'{i}.parquet' in os.listdir('/workspace/nemo_datasets/phu_classifier_filtering/'):
        continue
    df_result = pd.read_parquet(f'/workspace/nemo_datasets/result/{id_dict[i]}.parquet', engine='pyarrow')
    df_now = pd.read_parquet(f'/workspace/nemo_datasets/classifier_filtering/data/hq.parquet/part.{i}.parquet', engine='pyarrow')
    

    result = pd.merge(df_result, df_now, on=['id'], how='inner')
    
    result.rename(columns={'filename_x' : 'filename','text_x' : 'text'}, inplace=True)
    result.drop(['filename_y', 'text_y'], axis=1, inplace=True)
    #     result.to_parquet(f'/workspace/nemo_datasets/result/{i}.parquet', engine='pyarrow')
#     result.head()
    print(len(result))
    result.to_parquet(f'/workspace/nemo_datasets/phu_classifier_filtering/{i}.parquet', engine='pyarrow')
```
